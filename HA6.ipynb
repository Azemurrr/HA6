```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
from mpl_toolkits.mplot3d import Axes3D
```

Line plot

```python
x = [2, 3, 4, 5, 6]
y = [4, 9, 16, 25, 36]
plt.plot(x, y, marker='o')
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()
plt.show()
```

Bar plot

```python
labels = ['A', 'B', 'C', 'D']
values = [15, 36, 74, 34]

plt.bar(labels, values, color='skyblue')
plt.title("Bar Plot")
plt.ylabel("Values")
plt.show()
```

Scatter plot

3D plot

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

ax.plot_surface(X, Y, Z, cmap='viridis')
plt.title("3D Surface Plot")
plt.show()
```

Advanced plot

Box plot

```python
variables = np.random.normal(0, 1, 100)
sns.boxplot(data=variables)
plt.title("Box Plot")
plt.show()
```

2D Heatmap

```python
matrix = np.random.rand(5, 5)
sns.heatmap(matrix, annot=True, cmap='coolwarm')
plt.title("2D Heatmap")
plt.show()
```

Stacked Bar

```python
categories = ['A', 'B', 'C']
x = [5, 7, 6]
y = [3, 4, 5]

plt.bar(categories, x, label='Group 1')
plt.bar(categories, y, bottom=x, label='Group 2')
plt.title("Stacked Bar Plot")
plt.legend()
plt.show()
```

Gantt Chart

```python
df = pd.DataFrame({
    'Task': ['Task A', 'Task B', 'Task C'],
    'Start': [1, 3, 5],
    'Duration': [2, 4, 3]
})

plt.barh(df['Task'], df['Duration'], left=df['Start'])
plt.xlabel('Time')
plt.title("Gantt Chart")
plt.show()
```

Polar

```python
theta = np.linspace(0, 2 * np.pi, 100)
r = np.abs(np.sin(5 * theta))

plt.polar(theta, r)
plt.title("Polar Plot")
plt.show()
```
